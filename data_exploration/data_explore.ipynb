# Read in the Data

```python
import pandas as pd
import numpy as np
energy_df = pd.read_csv("../datasets/raw/eia_data.csv")
energy_df.head()
```

# Data Exploration

Filter down to all sectors only. Our analysis won't be focused on individual sectors

```python
# find unique sectors
energy_df["sectorDescription"].unique()
```

```python
# filter to just all sectors
energy_df = energy_df.loc[energy_df["sectorDescription"] == "All Sectors"]
energy_df.head()
```

Find the fuel types that are in the data and filters down to only the ones we are interested in

```python
energy_df["fuelTypeDescription"].unique()
```

```python
#filter to relevant fuel types. We can change this later
fuel_types = ["petroleum", "natural gas", "all coal products", "onshore wind turbine", "estimated total solar", "nuclear"]
energy_df = energy_df[energy_df["fuelTypeDescription"].isin(fuel_types)]
energy_df.head()
```

Now, we filter to the years we want (2015-2023 since that is what we have cost data for) 

```python
#filter to relevant fuel types. We can change this later
years = [2015 + x for x in range(9)]
energy_df = energy_df[energy_df["period"].isin(years)]
energy_df.head()
```

Checking now to see what the data looks like after that initial cleaning

```python
#row counts
energy_df.drop_duplicates(inplace=True)
energy_df.shape[0]
```

```python
#find columns with missing values
energy_df.isna().sum()
```

Here we remove the rows where generation is 0 or missing

```python
energy_df = energy_df[~(energy_df["generation"].isna() | (energy_df["generation"] == 0))]
energy_df.shape[0]
```

Start dropping some columns that aren't needed.

We aren't focused on breaking down consumption by UTO and Energy Generation, only total, so we can drop those.

We can drop the stocks, heat content, and reciepts columns because we don't care about physical quantities of material.

We can also remove the stateDescription, because the location is enough. We can also remove the sectorid because sectorName is enough.

We can also remove the physical unit columns, because BTU allows for better comparisons.

We can drop all the costs since most are missing and we have that data from seds and LBNL


```python
energy_df.drop(columns=["stateDescription", "sectorid", "consumption-for-eg", "consumption-for-eg-units", "consumption-for-eg-btu", "consumption-for-eg-btu-units", "consumption-uto", "consumption-uto-units", "consumption-uto-btu", "consumption-uto-btu-units", "total-consumption", "total-consumption-units", "heat-content", "heat-content-units", "receipts", "receipts-units", "receipts-btu", "receipts-btu-units", "stocks", "stocks-units", "cost", "cost-units", "cost-per-btu", "cost-per-btu-units"], inplace=True)
#rename location to "stateID" to match the other data
energy_df = energy_df.rename(columns={"location": "stateid"})
energy_df.isna().sum()
```

# Reading in Price Data

Some of the price data (Solar and wind) is already in price per MWh

The fossil fuel price data is in total expenditure, so we will need to make cost per MWh based on expenditure/generation.

```python
solar_price_df = pd.read_csv("../datasets/cleaned/solar_cost_per_state.csv")
solar_price_df.head()
```

```python
wind_price_df = pd.read_csv("../datasets/cleaned/wind_cost_per_state.csv")
wind_price_df.head()
```

I describe it more in the seds cleaning, but here each column is a seds datapoint, either cost statistic or co2 statistic for a certain type of fuel. 

The cost is in Millions of Dollars.

```python
seds_df = pd.read_csv("../datasets/cleaned/seds_cleaned.csv")
seds_df.head()
```

# Bringing the Data Together

Now we can add the solar and wind price per MWh and the raw cost for the othe fuels to the data.

```python
energy_df["Cost ($/MWh)"] = np.nan
solar_only = energy_df[energy_df["fuelTypeDescription"] == "estimated total solar"].copy()

solar_merged = solar_only.merge(
    solar_price_df.rename(columns={"State": "stateid", "Year": "period"}),
    on=["stateid", "period"],
    how="left"
)
energy_df.loc[energy_df["fuelTypeDescription"] == "estimated total solar", "Cost ($/MWh)"] = solar_merged["Price ($/MWh)"].values
```

```python
wind_only = energy_df[energy_df["fuelTypeDescription"] == "onshore wind turbine"].copy()

wind_merged = wind_only.merge(
    wind_price_df.rename(columns={"State": "stateid", "Year": "period"}),
    on=["stateid", "period"],
    how="left"
)

energy_df.loc[energy_df["fuelTypeDescription"] == "onshore wind turbine", "Cost ($/MWh)"] = wind_merged["Real 2024$/MWh"].values
```

Here we add the SEDS data of fuel type expenditure to a column called "raw cost" since this is in millions of dollars, not in $/MWh

We also add the CO2 emissions data

```python
energy_df["Raw Cost (Millions of $)"] = np.nan
energy_df["CO2 Emissions (Million Metric Tons)"] = np.nan
coal_only = energy_df[energy_df["fuelTypeDescription"] == "all coal products"].copy()
coal_merged = coal_only.merge(
    seds_df,
    on=["stateid", "period"],
    how="left"
)
energy_df.loc[energy_df["fuelTypeDescription"] == "all coal products", "Raw Cost (Millions of $)"] = coal_merged["CLEIV"].values
energy_df.loc[energy_df["fuelTypeDescription"] == "all coal products", "CO2 Emissions (Million Metric Tons)"] = coal_merged["CLTCE"].values

petroleum_only = energy_df[energy_df["fuelTypeDescription"] == "petroleum"].copy()
petroleum_merged = petroleum_only.merge(
    seds_df,
    on=["stateid", "period"],
    how="left"
)
energy_df.loc[energy_df["fuelTypeDescription"] == "petroleum", "Raw Cost (Millions of $)"] = petroleum_merged["PAEIV"].values
energy_df.loc[energy_df["fuelTypeDescription"] == "petroleum", "CO2 Emissions (Million Metric Tons)"] = petroleum_merged["PMTCE"].values

nuclear_only = energy_df[energy_df["fuelTypeDescription"] == "nuclear"].copy()
nuclear_merged = nuclear_only.merge(
    seds_df,
    on=["stateid", "period"],
    how="left"
)
energy_df.loc[energy_df["fuelTypeDescription"] == "nuclear", "Raw Cost (Millions of $)"] = nuclear_merged["NUEGV"].values

ng_only = energy_df[energy_df["fuelTypeDescription"] == "natural gas"].copy()
ng_merged = ng_only.merge(
    seds_df,
    on=["stateid", "period"],
    how="left"
)
energy_df.loc[energy_df["fuelTypeDescription"] == "natural gas", "Raw Cost (Millions of $)"] = ng_merged["NGEIV"].values
energy_df.loc[energy_df["fuelTypeDescription"] == "natural gas", "CO2 Emissions (Million Metric Tons)"] = ng_merged["NNTCE"].values
```

## Fixing Units in Data

We can now drop the units columns along with the sector description, since its the same value for all.

```python
energy_df.drop(["sectorDescription", "ash-content-units", "generation-units", "sulfur-content-units","total-consumption-btu-units"], 
               axis=1,
               inplace=True)
energy_df.head()
```

# Adding Derived Data

Read in the derived data

```python
ppmwh_df = pd.read_csv("../derived_data/price_per_mwh.csv")
ppmwh_df.head()
```

We aren't interested in generation or expenditure since its already in our data

```python
ppmwh_df.drop(columns=["Generation_MWh", "Expenditure_USD"], inplace=True)
ppmwh_df = ppmwh_df.rename(columns={"Price_per_MWh" : "Cost ($/MWh)"})
ppmwh_df.head()
```

```python
energy_df = energy_df.merge(
    ppmwh_df,
    on=["Year", "State", "fuelTypeDescription"],
    how="left"
)
#creates two cost columns, so we need to manually fill in missing with new derived data
energy_df["Cost ($/MWh)"] = energy_df["Cost ($/MWh)_x"].fillna(energy_df["Cost ($/MWh)_y"])
energy_df.drop(["Cost ($/MWh)_x", "Cost ($/MWh)_y"], axis=1, inplace=True)
```
